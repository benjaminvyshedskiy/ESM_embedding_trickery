```python
HIV_benchmark = pd.read_csv("/Users/benvyshedskiy/Documents/GitHub/ESM_embedding_trickery/QC/Well_Known_interactions/jager_2011_HIV.csv")
def process_benchmark(benchmark_dataset, filtered_data):
    possible_connection = 0
    interactions_had = 0
    for n in range(0,len(HIV_benchmark["vorf_fragment_ids"])):
        if type(benchmark_dataset["vorf_fragment_ids"][n])==str:
            possible_connection += len(benchmark_dataset["vorf_fragment_ids"][n].split(";"))
            for id in benchmark_dataset["vorf_fragment_ids"][n].split(";"):
                if id in list(filtered_data.columns) and benchmark_dataset["human_uniprot_id"][n] in list(filtered_data["prey"]):
                    if (list(filtered_data[filtered_data["prey"]==benchmark_dataset["human_uniprot_id"][n]][id])[0]==1):
                        interactions_had+=1
    return(possible_connection,interactions_had)

process_benchmark(HIV_benchmark,binary_interaction)
```
